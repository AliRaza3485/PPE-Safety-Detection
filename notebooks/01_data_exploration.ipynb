# PPE Safety Detection — Data Exploration

Quick sanity checks on the YOLO dataset before training:
class balance, sample images with bounding boxes, and image sizes.

```python
import sys
from pathlib import Path
from collections import Counter

# Make `src` importable when running from notebooks/
sys.path.append(str(Path.cwd().parent))
from src.utils import CLASS_NAMES, DATA_DIR

print('Classes:', CLASS_NAMES)
print('Data dir:', DATA_DIR)
```

```python
# Count instances per class across the training label files
counts = Counter()
label_dir = DATA_DIR / 'labels' / 'train'
for txt in label_dir.glob('*.txt'):
    for line in txt.read_text().splitlines():
        if line.strip():
            counts[int(line.split()[0])] += 1

for cls_id, n in sorted(counts.items()):
    print(f'{CLASS_NAMES[cls_id]:<16} {n}')
```
